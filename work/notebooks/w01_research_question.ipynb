## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

In this dataset (30,000 anonymized pages), the Spearman correlation between avg_position and ctr was -0.144 across all pages and -0.362 for pages with at least 100 impressions. Median CTR fell from 0.23 (positions 3-10) to 0.15 (10-20), 0.06 (20-50) and 0.00 (50+). Directionally, worse position goes with lower CTR, so any "low CTR" flag has to be adjusted for position.

```python
import pandas as pd, numpy as np
url = "https://raw.githubusercontent.com/SamyElnagar11/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

sub = df[df["impressions_90d"] >= 100].copy()
print("Pages:", len(df), "| pages with >=100 impressions:", len(sub))
print("Spearman avg_position vs ctr (all pages):", round(df["avg_position"].corr(df["ctr"], method="spearman"), 3))
print("Spearman avg_position vs ctr (>=100 impressions):", round(sub["avg_position"].corr(sub["ctr"], method="spearman"), 3))

sub["pos_bucket"] = pd.cut(sub["avg_position"], [0, 3, 10, 20, 50, np.inf], include_lowest=True)
print(sub.groupby("pos_bucket", observed=True)["ctr"].agg(["count", "median"]).round(3))
```
