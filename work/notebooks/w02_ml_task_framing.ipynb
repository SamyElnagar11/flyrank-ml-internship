## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Proxy target:** `ctr_gap` = (median CTR of the page's position tier) minus (the page's own CTR). A positive gap means the page gets fewer clicks than similar-position pages.

This is a proxy, not a proven "bad title" label: low CTR can have other causes (search-result layout, intent mismatch, brand queries). It is computed from the same window as the features, so it is a starting point, not a future outcome. A stronger version would predict CTR over a later window.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

**Unit of analysis:** one row = one page (anonymized). The table below is my lane slice (pages with at least 100 impressions in 90 days) with the sketched target column `ctr_gap`.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd, numpy as np
url = "https://raw.githubusercontent.com/SamyElnagar11/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)


lane = df[(df["impressions_90d"] >= 100) & (df["avg_position"] > 0)].copy()


tier_median = lane.groupby("position_tier")["ctr"].transform("median")
lane["ctr_gap"] = tier_median - lane["ctr"]

cols = ["impressions_90d", "avg_position", "position_tier", "ctr", "ctr_gap"]
print("One row = one page. Rows in lane slice:", len(lane))
lane[cols].sort_values("ctr_gap", ascending=False).head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as "CTR below 0.5 and position 1-20" treats every position the same and needs a hand-picked threshold. In this dataset I observed that median CTR falls as position gets worse, so one cutoff is unfair to some tiers. A scoring approach compares each page with its own tier, and a model could additionally use signals like impressions, intent, content type, and age. It is only worth it if it beats the tier-median baseline on unseen clients. If it does not, the simple baseline is the honest answer.
